# **QUESTION PRACTICE OF NUMPY**

Check wheather a sudoku is valid or not

**METHOD - 1**

```python
import numpy as np
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 3, 5, 7, 9, 1],
    [7, 1, 3, 9, 4, 2, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
valid = "true"
set_no = set(range(1,10))
row_sum = np.sum(s,axis = 1)
col_sum = np.sum(s,axis = 0)
for i in row_sum :
  if i != 45 :
    valid = "false"
    break
for y in col_sum :
  if y != 45 :
    valid = "false"
    break
for z in range(0,9,3):
  if s[z:z+3,0:3].sum() != 45:
    valid = "false"
  if s[z:z+3,3:6].sum() != 45 :
    valid = "false"
  if s[z:z+3,6:9].sum() != 45:
    valid = "false"
for i in s :
  ns = set()
  for y in i :
     if y != 0 :
      if y not in ns:
       ns.add(y)
  if len(ns) != 9 :
    valid = 'false'
for d in s :
  if set(d) != set_no :
    valid = "false"
for x in range(0,9):
  col = s[:,x]
  if set(col) != set_no :
     valid = "false"
if valid == 'true':
  print("valid sudoku")
else :
  print("invalid sudoku")

```
