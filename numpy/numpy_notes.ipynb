DEEP AND SHALLOW COPY

shallow copy is when you slice out something from array and change in that slice would not affect the main array
and in deep copy its opposite if you change in slice it will got change in main array

```python
a = np.arange(1,11)
slice = a[1:7]
print(a)
print(slice)
slice = slice*10
print(slice)
print(a)#no change in main array so shallow copy
```

```python
a = np.array([1, 2, 3, 4, 5])

b = a[1:4]

b[0] = 100

print(a)
print(b)
```

```python
a = np.arange(1,11)
print(a)
b = a.copy()
b[0] = 27
print(a)
print(b) # this time no change in a because we use .copy()it converted into shallow copy
```

ADVANCED ARRAY MANIPULATION


stacking array

you can't stack matrix it only done on array simple

```python
#vstack = vertical stack
a = np.arange(1,6)
b = np.arange(6,11)
print(a)
print(b)
print(np.vstack((a,b)))
# this is called vertical stacking
```

```python
#horizontal stacking
hor = np.hstack((a,b))
print(hor)
#simply converted into 1 dimension by adding horizontally
```

```python
#column stack
col = np.column_stack((a,b))
col
```

splitting array

```python
#hsplit means horizontal spilitiing
np.hsplit(col,2) # split col into 2
```

```python
np.vsplit(col,5)
```
